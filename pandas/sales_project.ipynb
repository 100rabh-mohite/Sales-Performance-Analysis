```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/Rohit/Downloads/sales_dataset.csv")
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
# Analyze total sales per customer.
customer_sales=df.groupby("CustomerID")["Sales"].sum()
print(customer_sales)
```

```python
#Calculate moving average sales per month.
df["Moving_Avg"]=df["Sales"].rolling(window=3).mean()
```

```python
#Segment customers based on total spending (e.g., low, medium, high).
def segment(x):
    if x < 1000:
        return "Low"
    elif x < 5000:
        return "Medium"
    else:
        return "High"

customer_total = df.groupby("CustomerID")["Sales"].sum()

segments = customer_total.apply(segment)

print(segments)
```

```python
#Identify top 10 customers by lifetime value.
top_customers = (df.groupby("CustomerID")["Sales"].sum().sort_values(ascending=False).head(10))
print(top_customers)
```
